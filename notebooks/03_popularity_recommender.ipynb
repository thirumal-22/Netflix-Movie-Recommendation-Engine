```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/cleaned_netflix_ratings.csv")

movie_stats = df.groupby(["MovieID", "Title", "Genre", "Year"]).agg(
    avg_rating=("Rating", "mean"),
    rating_count=("Rating", "count")
).reset_index()

movie_stats["popularity_score"] = movie_stats["avg_rating"] * np.log1p(movie_stats["rating_count"])

top_movies = movie_stats.sort_values("popularity_score", ascending=False).head(10)

print(top_movies[["Title", "Genre", "Year", "avg_rating", "rating_count", "popularity_score"]])
```
